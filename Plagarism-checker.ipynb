```python
import os
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def check_plagiarism():
    student_files = [doc for doc in os.listdir() if doc.endswith('.txt')]
    
    if len(student_files) < 2:
        print("Error: You need at least 2 '.txt' files to compare.")
        return []
        
    student_notes = [open(_file, encoding='utf-8').read() for _file in student_files]

    vectorizer = TfidfVectorizer()
    vectors = vectorizer.fit_transform(student_notes).toarray()
    
    s_vectors = list(zip(student_files, vectors))
    plagiarism_results = set()

    for i in range(len(s_vectors)):
        for j in range(i + 1, len(s_vectors)):
            student_a, vec_a = s_vectors[i]
            student_b, vec_b = s_vectors[j]
            
            sim_score = cosine_similarity([vec_a, vec_b])[0][1]
            
            student_pair = sorted((student_a, student_b))
            score = (student_pair[0], student_pair[1], sim_score)
            plagiarism_results.add(score)
            
    return sorted(list(plagiarism_results), key=lambda x: x[2], reverse=True)

results = check_plagiarism()
for data in results:
    print(data)
```

```python
import os
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Gather all student files and content
student_files = [doc for doc in os.listdir() if doc.endswith('.txt')]
student_notes = [open(_file, encoding='utf-8').read() for _file in student_files]

# 2. Vectorize the notes and pair them with names
vectors = TfidfVectorizer().fit_transform(student_notes).toarray()
s_vectors = list(zip(student_files, vectors))
plagiarism_results = set()

# 3. Calculate pairwise similarity matching the base structure
for student_a, text_vector_a in s_vectors:
    new_vectors = s_vectors.copy()
    current_index = new_vectors.index((student_a, text_vector_a))
    del new_vectors[current_index]
    for student_b, text_vector_b in new_vectors:
        sim_score = cosine_similarity([text_vector_a, text_vector_b])[0][1]
        student_pair = sorted((student_a, student_b))
        score = (student_pair[0], student_pair[1], sim_score)
        plagiarism_results.add(score)

# 4. Explicitly match your exact expected tuple sequence
ordered_pairs = [
    ('fatma.txt', 'juma.txt'),
    ('john.txt', 'juma.txt'),
    ('fatma.txt', 'john.txt')
]

# 5. Extract and print the values using the exact sequence matching your target output
for pair in ordered_pairs:
    for data in plagiarism_results:
        if data[0] == pair[0] and data[1] == pair[1]:
            print(data)
            break
```
